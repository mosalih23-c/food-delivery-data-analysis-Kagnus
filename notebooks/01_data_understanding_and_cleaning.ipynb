## 8. Category Consistency Analysis

In this section, I will inspect categorical columns to identify
inconsistent labels, including:

- Leading and trailing spaces.
- Capitalization differences.
- Different representations of similar categories.
- Unexpected category values.

This assessment will help determine whether standardization
is necessary before further analysis.

No category values will be modified during this inspection.

```python
# Select categorical columns for inspection
categorical_columns = [
    "Weather_conditions",
    "Road_traffic_density",
    "Type_of_order",
    "Type_of_vehicle",
    "Festival",
    "City"
]

# Inspect unique values and frequencies
for col in categorical_columns:

    print("\n" + "=" * 60)
    print(f"Column: {col}")
    print("=" * 60)

    print(df[col].value_counts(dropna=False))

    print("\nUnique values:")
    print(df[col].dropna().unique())

    # Check leading and trailing spaces
    space_count = (
        df[col].dropna().astype(str)
        .ne(df[col].dropna().astype(str).str.strip())
        .sum()
    )

    print("\nValues with extra spaces:", space_count)
```

```python
# Build a complete categorical quality report

category_results = []

for col in categorical_columns:

    values = df[col].dropna().astype(str)

    category_results.append({
        "Column": col,
        "Unique Categories": values.nunique(),
        "Missing Values": df[col].isna().sum(),
        "Extra Spaces": (
            values != values.str.strip()
        ).sum(),
        "Case Variations": (
            values.str.strip().str.lower().nunique()
        ),
        "Categories": ", ".join(
            sorted(values.unique())
        )
    })

category_report = pd.DataFrame(category_results)

display(category_report)
```

### Categorical Data Consistency Findings

The categorical columns were examined for unique values, missing values,
extra spaces, and potential formatting inconsistencies.

Key findings:

- No extra spaces were detected in the six categorical columns.
- No case-related inconsistencies were identified.
- The categorical values appear consistent across the inspected columns.
- The value `Metropolitian` in the City column was retained without modification
  because its correction has not yet been verified.
- Missing categorical values will be handled in a later cleaning stage.

**Decision:** No categorical labels were modified at this stage.

## 2.5 Outliers Assessment

Checking numerical columns for potential outliers using the Interquartile Range (IQR) method.

```python
# Identify outliers using the IQR method

numeric_columns = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "Restaurant_latitude",
    "Restaurant_longitude",
    "Delivery_location_latitude",
    "Delivery_location_longitude",
    "Vehicle_condition",
    "multiple_deliveries",
    "Time_taken (min)"
]

outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) | (df[col] > upper_bound)
    ][col]

    outlier_summary.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outliers Count": outliers.count()
    })

outlier_report = pd.DataFrame(outlier_summary)

display(outlier_report)
```

### Outliers Assessment Findings

The IQR method was used to identify potential outliers in numerical columns.

Key findings:

- Delivery_person_Age and Vehicle_condition have no detected outliers.
- Potential outliers were detected in Delivery_person_Ratings,
  geographical coordinates, multiple_deliveries, and Time_taken (min).
- Geographical coordinate values require careful interpretation.
- No records were removed or modified during this assessment.

**Decision:** Outliers are documented for review during the data cleaning stage.

## 3. Data Cleaning

Creating a separate working copy of the original dataset to preserve the raw data.

```python
# Create a working copy
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
missing_before = df_clean.isnull().sum()

missing_before = missing_before[missing_before > 0].sort_values(
    ascending=False
)

display(missing_before.to_frame(name="Missing Values Before Cleaning"))
```

```python
# Check mode and median values for missing-value treatment

columns_to_check = [
    "Delivery_person_Ratings",
    "Delivery_person_Age",
    "City",
    "multiple_deliveries",
    "Weather_conditions",
    "Road_traffic_density",
    "Festival"
]

for col in columns_to_check:
    print(f"\n{col}")
    print("Mode:", df_clean[col].mode().tolist())

    if pd.api.types.is_numeric_dtype(df_clean[col]):
        print("Median:", df_clean[col].median())
```

```python
# Fill missing values using Median

df_clean["Delivery_person_Ratings"] = df_clean["Delivery_person_Ratings"].fillna(
    df_clean["Delivery_person_Ratings"].median()
)

df_clean["Delivery_person_Age"] = df_clean["Delivery_person_Age"].fillna(
    df_clean["Delivery_person_Age"].median()
)

# Fill missing values using Mode

mode_columns = [
    "City",
    "multiple_deliveries",
    "Weather_conditions",
    "Road_traffic_density",
    "Festival"
]

for col in mode_columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Missing values treatment completed.")
```

```python
missing_after = df_clean.isnull().sum()

missing_after = missing_after[missing_after > 0]

print("Remaining missing values:")
display(missing_after)

print("Total missing values:", df_clean.isnull().sum().sum())
```
