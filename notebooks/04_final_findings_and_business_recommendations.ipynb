# Food Delivery Data Analysis – Final Findings

## 1. Final Key Findings

### Finding 1: Delivery Time Baseline

**Observation:** Delivery duration is generally centered around 26 minutes, with a slightly right-skewed distribution.

**Numerical Evidence:**
- Mean delivery time: 26.29 minutes
- Median delivery time: 26 minutes
- Minimum delivery time: 10 minutes
- Maximum delivery time: 54 minutes
- Standard deviation: 9.38 minutes

**Operational Interpretation:**
Most deliveries are completed around the 26-minute mark. However, the variability in delivery duration indicates that delivery performance is not consistent across all orders.

---

### Finding 2: Traffic Congestion Is Associated with Longer Delivery Times

**Observation:** Deliveries under Jam traffic conditions take considerably longer than deliveries under Low traffic conditions.

**Numerical Evidence:**
- Jam traffic mean: 31.18 minutes (14,139 orders)
- Low traffic mean: 21.46 minutes (16,077 orders)
- Difference: 9.72 minutes

**Operational Interpretation:**
Traffic congestion is an important factor associated with delivery delays. Delivery planning and rider allocation during peak congestion may help improve operational efficiency.

---

### Finding 3: Delivery Duration Varies Across Weather Conditions

**Observation:** Average delivery duration differs noticeably across weather conditions.

**Numerical Evidence:**
- Cloudy: 28.92 minutes
- Fog: 28.74 minutes
- Sunny: 21.86 minutes
- Difference between Cloudy and Sunny: 7.06 minutes

**Operational Interpretation:**
Weather conditions may influence delivery efficiency. Weather-related factors should be considered when estimating delivery times and planning daily operations.

---

### Finding 4: Multiple Deliveries Increase Delivery Duration

**Observation:** Delivery duration increases considerably as the number of multiple deliveries assigned to a delivery person increases.

**Numerical Evidence:**

| Multiple Deliveries | Mean Duration | Orders |
|---|---:|---:|
| 0 | 22.88 minutes | 14,094 |
| 1 | 26.72 minutes | 29,144 |
| 2 | 40.45 minutes | 1,985 |
| 3 | 47.82 minutes | 361 |

The difference between 0 and 3 multiple deliveries is 24.94 minutes.

**Operational Interpretation:**
Assigning several deliveries to the same delivery person may substantially increase customer waiting time. Delivery batching policies should balance efficiency with delivery speed.

---

### Finding 5: Delivery Duration Differs Across Vehicle Types

**Observation:** Average delivery duration varies across vehicle types.

**Numerical Evidence:**
- Motorcycle: 27.61 minutes (26,429 orders)
- Scooter: 24.48 minutes (15,273 orders)
- Electric Scooter: 24.47 minutes (3,814 orders)
- Bicycle: 26.43 minutes (68 orders)

**Operational Interpretation:**
Vehicle type may be associated with delivery performance. However, these differences may also be affected by distance, traffic, and order allocation patterns.

---

### Finding 6: Delivery Distance Has a Positive Relationship with Delivery Duration

**Observation:** Longer delivery distances tend to be associated with longer delivery durations, although the relationship is not strong.

**Numerical Evidence:**
- Average distance: 9.72 km
- Median distance: 9.19 km
- Pearson correlation: 0.321
- Records included in distance analysis: 41,513

**Operational Interpretation:**
Distance contributes to delivery duration, but it is not the only influencing factor. Traffic, weather, and multiple deliveries may also affect delivery performance.

---

### Finding 7: Delivery Ratings Are Negatively Associated with Delivery Duration

**Observation:** Lower delivery ratings tend to be associated with longer delivery durations.

**Numerical Evidence:**
- Pearson correlation between ratings and delivery duration: −0.331
- Average duration for ratings around 3.5–3.9: approximately 37 minutes
- Average duration for ratings around 4.5–4.9: approximately 24 minutes

**Operational Interpretation:**
Longer delivery times may be associated with lower customer ratings, suggesting that delivery speed may be relevant to customer experience.

**Limitation:** Missing ratings were imputed using the median, and the dataset contains rating values of 6. Therefore, rating-related findings should be interpreted cautiously.
